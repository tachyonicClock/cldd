```python
import optuna
import matplotlib.pyplot as plt
import seaborn as sns

storage = optuna.storages.JournalStorage(
    optuna.storages.JournalFileStorage("../logs/optuna.journal.log")
)

study_name = "bocl/tune_strategy/abrupt/LWF/oracle"
study = optuna.load_study(study_name=study_name, storage=storage)
df = study.trials_dataframe()
df.head()
```

```python
sns.regplot(
    data=df,
    x="params_learner.alpha",
    y="user_attrs_accuracy_final",
    scatter=True,
    lowess=True,
)
plt.title(study_name)
# plt.xscale("log")
```
