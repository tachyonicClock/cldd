```python
import pandas as pd
import numpy as np

df = pd.read_csv("logs/dd_run/data_frame.csv")
df
# replace inf with NaN
df.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
# Aggregation Step
AGGREGATIONS = ("mean",)

df_grouped = df.groupby(["boundary", "detector"]).aggregate(
    {
        "dd.f1": AGGREGATIONS,
        "dd.wasserstein_distance": AGGREGATIONS,
    }
)
# flatten the multi-index columns
# df_grouped.columns = ['_'.join(col) for col in df_grouped.columns]
# df_grouped
```

```python
# Styling Step
def style_max(s, props):
    return np.where(s == np.nanmax(s.values), props, "")


def style_min(s, props):
    return np.where(s == np.nanmin(s.values), props, "")


(
    df_grouped.style.format(precision=3)
    .apply(style_max, props="font-weight: bold;", subset=["dd.f1"])
    .apply(style_min, props="font-weight: bold;", subset=["dd.wasserstein_distance"])
)
```
