```python
import torch
import matplotlib.pyplot as plt
from torch import Tensor
from typing import Sequence
from torchvision.utils import make_grid


dataset = torch.load("data/rotated_fashion_mnist_data.pt")
train: Sequence[Tensor] = dataset["train"]
```

```python
ratio = 2 + 2**0.5
width = 390  # pt
height = width / ratio
grid_len = 3

# Use serif font
plt.rcParams["font.family"] = "serif"

fig, axes = plt.subplots(1, 5, figsize=(width / 72, height / 72))

# For each task
for i in range(5):
    print()
    grid = 255 - make_grid(
        train[i][0][: int(grid_len * grid_len)], nrow=grid_len, padding=2
    )
    axes[i].imshow(grid[0], cmap="gray")
    axes[i].set_title(f"Task {i + 1}")
    axes[i].axis("off")


# Save the figure. No whitespace around the figure, and tight layout.
plt.savefig("data/rotated_fashion_mnist_tasks.pdf", bbox_inches="tight", pad_inches=0)
```
