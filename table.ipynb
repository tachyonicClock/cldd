```python
import pandas as pd

df = pd.read_csv("logs/evaluate/data_frame.csv")
```

```python
df_grouped = df.groupby(["detector_label", "strategy", "boundary"]).aggregate(
    {
        "ocl.accuracy_all_avg": ("mean", "std"),
        "ocl.accuracy_seen_avg": ("mean", "std"),
        "ocl.forward_transfer": ("mean", "std"),
        "ocl.backward_transfer": ("mean", "std"),
        "dd.wasserstein_distance": ("mean", "std"),
    }
)
```

```python
# Apply per numeric column colour gradient
df_grouped.style.background_gradient(cmap="viridis")
```
